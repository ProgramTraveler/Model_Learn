```python
import torch

# 1. 创建需要训练的参数
x = torch.tensor(1.0, requires_grad=True)

learning_rate = 0.1

for step in range(10):
    # 2. 使用当前 x 计算损失
    loss = (x - 3) ** 2

    # 3. 清空上一轮的梯度
    if x.grad is not None:
        x.grad.zero_()

    # 4. 计算当前梯度
    loss.backward()

    # 5. 更新参数
    with torch.no_grad():
        x -= learning_rate * x.grad

    print(
        "轮数：", step + 1,
        "x：", x.item(),
        "loss：", loss.item()
    )
```

```python
import torch

# 训练数据，规律是 y = 2x + 1
inputs = torch.tensor([1.0, 2.0, 3.0, 4.0])
targets = torch.tensor([3.0, 5.0, 7.0, 9.0])

# 模型需要学习的两个参数
w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

learning_rate = 0.01

# 训练
for step in range(500):
    predictions = w * inputs + b
    loss = ((predictions - targets) ** 2).mean()

    if w.grad is not None:
        w.grad.zero_()
        b.grad.zero_()

    loss.backward()

    with torch.no_grad():
        w -= learning_rate * w.grad
        b -= learning_rate * b.grad

# 使用最终参数重新计算
with torch.no_grad():
    final_predictions = w * inputs + b
    final_loss = ((final_predictions - targets) ** 2).mean()

print("PyTorch 版本：", torch.__version__)
print("最终 w：", w.item())
print("最终 b：", b.item())
print("最终预测：", final_predictions)
print("最终 loss：", final_loss.item())
```

```python
import torch
from torch import nn

# 固定随机初始化，确保每次结果一致
torch.manual_seed(42)

# 训练数据：真实规律是 y = 2x + 1
inputs = torch.tensor([
    [1.0],
    [2.0],
    [3.0],
    [4.0]
])

targets = torch.tensor([
    [3.0],
    [5.0],
    [7.0],
    [9.0]
])

# 自动创建 weight 和 bias
model = nn.Linear(
    in_features=1,
    out_features=1,
    bias=True
)

# 平均平方误差
loss_function = nn.MSELoss()

# 训练前的预测和 loss
predictions = model(inputs)
loss = loss_function(predictions, targets)

print("模型：", model)
print("输入形状：", inputs.shape)
print("初始预测：")
print(predictions)
print("初始 loss：", loss.item())

for name, parameter in model.named_parameters():
    print(name, parameter)
```

```python
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)

print(optimizer)
```

```python
model.parameters()
```

```python
# 1. 清空模型所有参数的旧梯度
optimizer.zero_grad()

# 2. 使用当前模型计算预测和 loss
predictions = model(inputs)
loss = loss_function(predictions, targets)

# 3. 反向传播，计算 weight 和 bias 的梯度
loss.backward()

print("weight 的梯度：", model.weight.grad)
print("bias 的梯度：", model.bias.grad)

# 4. 根据梯度更新 weight 和 bias
optimizer.step()
```

```python
for step in range(500):
    # 清空旧梯度
    optimizer.zero_grad()

    # 前向传播
    predictions = model(inputs)

    # 计算损失
    loss = loss_function(predictions, targets)

    # 反向传播
    loss.backward()

    # 更新模型参数
    optimizer.step()

    if (step + 1) % 100 == 0:
        print(
            "轮数：", step + 1,
            "loss：", loss.item()
        )
```

```python
import torch
from torch import nn

# 固定随机数，保证模型初始参数一致
torch.manual_seed(42)

# 训练数据
inputs = torch.tensor([
    [1.0, 1.0],
    [2.0, 1.0],
    [1.0, 2.0],
    [2.0, 2.0]
])

# 正确答案来自 y = 2x₁ + 3x₂ + 1
targets = torch.tensor([
    [6.0],
    [8.0],
    [9.0],
    [11.0]
])

# 多层模型
model = nn.Sequential(
    nn.Linear(2, 3),
    nn.ReLU(),
    nn.Linear(3, 1)
)

# 损失函数和优化器
loss_function = nn.MSELoss()

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)

# 训练前的 loss
with torch.no_grad():
    initial_predictions = model(inputs)
    initial_loss = loss_function(initial_predictions, targets)

print("训练前 loss：", initial_loss.item())

# 完成一轮训练
optimizer.zero_grad()

predictions = model(inputs)
loss = loss_function(predictions, targets)

loss.backward()
optimizer.step()

# 检查更新后的 loss
with torch.no_grad():
    new_predictions = model(inputs)
    new_loss = loss_function(new_predictions, targets)

print("更新后 loss：", new_loss.item())
print("更新后的预测：")
print(new_predictions)
```

```python
for epoch in range(1000):
    # 进入训练模式
    model.train()

    # 清空旧梯度
    optimizer.zero_grad()

    # 前向传播
    predictions = model(inputs)

    # 计算损失
    loss = loss_function(predictions, targets)

    # 反向传播
    loss.backward()

    # 更新所有层的参数
    optimizer.step()

    if (epoch + 1) % 100 == 0:
        print(
            "训练轮数：", epoch + 1,
            "loss：", loss.item()
        )
```

```python
model.eval()

with torch.no_grad():
    final_predictions = model(inputs)
    final_loss = loss_function(final_predictions, targets)

print("最终预测：")
print(final_predictions)

print("正确答案：")
print(targets)

print("最终 loss：", final_loss.item())
```

```python
def train_and_evaluate(hidden_size):
    torch.manual_seed(42)

    candidate_model = nn.Sequential(
        nn.Linear(1, hidden_size),
        nn.ReLU(),
        nn.Linear(hidden_size, 1)
    )

    candidate_optimizer = torch.optim.SGD(
        candidate_model.parameters(),
        lr=0.01
    )

    for epoch in range(3000):
        candidate_model.train()

        candidate_optimizer.zero_grad()

        predictions = candidate_model(
            nonlinear_inputs
        )

        loss = loss_function(
            predictions,
            nonlinear_targets
        )

        loss.backward()
        candidate_optimizer.step()

    candidate_model.eval()

    with torch.no_grad():
        train_predictions = candidate_model(
            nonlinear_inputs
        )

        train_loss = loss_function(
            train_predictions,
            nonlinear_targets
        )

        validation_predictions = candidate_model(
            validation_inputs
        )

        validation_loss = loss_function(
            validation_predictions,
            validation_targets
        )

    return (
        candidate_model,
        train_loss.item(),
        validation_loss.item()
    )
```

```python
for hidden_size in [2, 4, 8, 16]:
    _, train_loss, validation_loss = train_and_evaluate(
        hidden_size
    )

    print(
        "隐藏神经元：", hidden_size,
        "训练 loss：", train_loss,
        "验证 loss：", validation_loss
    )
```

```python
# 导入 PyTorch
import torch

# 从 PyTorch 中导入神经网络模块
from torch import nn


# 固定随机种子，
# 确保模型每次运行时使用相同的初始参数
torch.manual_seed(42)


# 创建训练输入
# 一共有 6 条数据，每条数据有 2 个特征
classification_inputs = torch.tensor([
    [-2.0, -1.0],
    [-1.0, -2.0],
    [-1.0, -1.0],
    [1.0, 1.0],
    [2.0, 1.0],
    [1.0, 2.0]
])


# 创建每条数据对应的正确类别
# 前三条属于类别 0
# 后三条属于类别 1
classification_targets = torch.tensor([
    0,
    0,
    0,
    1,
    1,
    1
])


# 创建二分类模型
classifier = nn.Sequential(
    # 每条数据有 2 个输入特征，
    # 第一层产生 4 个隐藏特征
    nn.Linear(
        in_features=2,
        out_features=4
    ),

    # 把负数变成 0，
    # 为模型增加非线性表达能力
    nn.ReLU(),

    # 接收 4 个隐藏特征，
    # 输出类别 0 和类别 1 的两个分数
    nn.Linear(
        in_features=4,
        out_features=2
    )
)


# 创建交叉熵损失函数
# 它用于比较模型输出的类别分数与正确类别
classification_loss_function = nn.CrossEntropyLoss()


# 创建随机梯度下降优化器
# model.parameters() 包含模型中所有 weight 和 bias
classification_optimizer = torch.optim.SGD(
    classifier.parameters(),
    lr=0.1
)


# 设置训练轮数
number_of_epochs = 200


# 开始训练循环
for epoch in range(number_of_epochs):
    # 把模型切换到训练模式
    classifier.train()

    # 清空上一轮保留的所有梯度
    classification_optimizer.zero_grad()

    # 前向传播：
    # 计算每条数据属于两个类别的分数
    scores = classifier(
        classification_inputs
    )

    # 使用类别分数和正确标签计算 loss
    loss = classification_loss_function(
        scores,
        classification_targets
    )

    # 反向传播：
    # 计算模型中所有参数的梯度
    loss.backward()

    # 根据梯度和学习率更新所有模型参数
    classification_optimizer.step()

    # 每训练 50 轮显示一次 loss
    if (epoch + 1) % 50 == 0:
        print(
            "训练轮数：", epoch + 1,
            "loss：", loss.item()
        )


# 把模型切换到评估模式
classifier.eval()


# 评估阶段不需要计算梯度
with torch.no_grad():
    # 使用训练好的模型重新计算类别分数
    final_scores = classifier(
        classification_inputs
    )

    # 对每一行类别分数，
    # 选择分数最大的位置作为预测类别
    predicted_classes = final_scores.argmax(
        dim=1
    )

    # 比较预测类别和正确类别
    correct_results = (
        predicted_classes == classification_targets
    )

    # True 会转换成 1，False 会转换成 0
    # 然后计算平均值得到准确率
    accuracy = correct_results.float().mean()


# 显示每条数据的两个类别分数
print("\n最终类别分数：")
print(final_scores)


# 显示模型预测的类别
print("\n模型预测：")
print(predicted_classes)


# 显示正确类别
print("\n正确答案：")
print(classification_targets)




# 显示每条数据是否预测正确
print("\n是否预测正确：")
print(correct_results)


# 显示最终准确率
print("\n准确率：", accuracy.item())
```

```python
# 导入 PyTorch
import torch

# 导入神经网络工具
from torch import nn

# 固定随机数，使每次运行的初始参数和结果尽量一致
torch.manual_seed(42)


# =========================
# 1. 准备训练数据
# =========================

# 每个样本有两个输入特征
classification_inputs = torch.tensor([
    [-2.0, -1.0],
    [-1.0, -2.0],
    [-1.0, -1.0],
    [ 1.0,  1.0],
    [ 2.0,  1.0],
    [ 1.0,  2.0]
])

# 正确类别：
# 前三个样本属于类别 0
# 后三个样本属于类别 1
classification_targets = torch.tensor([
    0, 0, 0, 1, 1, 1
])


# =========================
# 2. 创建分类模型
# =========================

classifier = nn.Sequential(
    # 每个样本有 2 个特征，隐藏层包含 4 个神经元
    nn.Linear(2, 4),

    # 加入非线性能力
    nn.ReLU(),

    # 输出两个数，分别代表类别 0 和类别 1 的分数
    nn.Linear(4, 2)
)


# =========================
# 3. 创建损失函数和优化器
# =========================

# 交叉熵损失用于分类问题
loss_function = nn.CrossEntropyLoss()

# 优化器负责更新模型中的所有权重和偏置
optimizer = torch.optim.SGD(
    classifier.parameters(),
    lr=0.1
)


# =========================
# 4. 训练模型
# =========================

classifier.train()

for epoch in range(200):
    # 使用当前参数计算分类分数
    scores = classifier(classification_inputs)

    # 比较分类分数和正确类别，计算 loss
    loss = loss_function(scores, classification_targets)

    # 清除上一轮留下的梯度
    optimizer.zero_grad()

    # 反向传播，计算每个参数的梯度
    loss.backward()

    # 根据梯度更新模型参数
    optimizer.step()


# =========================
# 5. 检查训练结果
# =========================

# 切换到评估模式
classifier.eval()

# 评估时不需要记录计算图和梯度
with torch.no_grad():
    # 得到每个样本对应两个类别的分数
    final_scores = classifier(classification_inputs)

    # 在两个类别分数中，选择分数最大的位置
    predictions = torch.argmax(final_scores, dim=1)

    # 判断预测类别是否和正确类别一致
    correct = predictions == classification_targets

    # 计算准确率
    accuracy = correct.float().mean()

print("类别分数：")
print(final_scores)

print("\n预测类别：", predictions)
print("正确类别：", classification_targets)
print("准确率：", accuracy.item())
```

```python
# =========================
# 13. 准备三个类别的训练数据
# =========================

# 每一行仍然是一个样本
# 每个样本仍然包含两个输入特征
three_class_inputs = torch.tensor([
    # 类别 0：主要位于左下方
    [-2.0, -2.0],
    [-2.0, -1.0],
    [-1.0, -2.0],

    # 类别 1：主要位于右下方
    [ 2.0, -2.0],
    [ 2.0, -1.0],
    [ 1.0, -2.0],

    # 类别 2：主要位于上方
    [ 0.0,  2.0],
    [-0.5,  1.0],
    [ 0.5,  1.0]
])

# 为每个样本提供正确类别
three_class_targets = torch.tensor([
    0, 0, 0,   # 前三个属于类别 0
    1, 1, 1,   # 中间三个属于类别 1
    2, 2, 2    # 最后三个属于类别 2
])

print("输入数据：")
print(three_class_inputs)

print("\n正确类别：")
print(three_class_targets)

print("\n输入形状：", three_class_inputs.shape)
print("正确类别形状：", three_class_targets.shape)
```

```python
# =========================
# 14. 创建三分类模型
# =========================

three_class_model = nn.Sequential(
    # 每个样本包含两个输入特征
    # 隐藏层设置为 8 个神经元
    nn.Linear(2, 8),

    # 加入非线性能力
    nn.ReLU(),

    # 最终输出三个类别分数
    nn.Linear(8, 3)
)

print(three_class_model)
```

```python
# =========================
# 15. 创建损失函数和优化器
# =========================

# 创建交叉熵损失函数
# 它可以处理两个或多个类别
three_class_loss_function = nn.CrossEntropyLoss()

# 创建优化器
# 它负责更新三分类模型中的权重和偏置
three_class_optimizer = torch.optim.SGD(
    three_class_model.parameters(),
    lr=0.1
)


# =========================
# 16. 训练三分类模型
# =========================

# 切换到训练模式
three_class_model.train()

# 训练 300 轮
for epoch in range(300):
    # 第一步：清除上一轮留下的梯度
    three_class_optimizer.zero_grad()

    # 第二步：计算每个样本的三个类别分数
    three_class_scores = three_class_model(
        three_class_inputs
    )

    # 第三步：比较类别分数与正确类别，计算 loss
    three_class_loss = three_class_loss_function(
        three_class_scores,
        three_class_targets
    )

    # 第四步：反向传播，计算模型参数的梯度
    three_class_loss.backward()

    # 第五步：根据梯度更新模型参数
    three_class_optimizer.step()

    # 每训练 50 轮，显示一次当前 loss
    if (epoch + 1) % 50 == 0:
        print(
            "训练轮数：",
            epoch + 1,
            "loss：",
            three_class_loss.item()
        )
```

```python
# =========================
# 17. 检查三分类训练结果
# =========================

# 切换到评估模式
three_class_model.eval()

# 评估时不需要计算梯度
with torch.no_grad():
    # 得到每个训练样本的三个类别分数
    final_three_class_scores = three_class_model(
        three_class_inputs
    )

    # 把三个类别分数转换成概率
    final_three_class_probabilities = torch.softmax(
        final_three_class_scores,
        dim=1
    )

    # 选择每个样本概率最高的类别
    final_three_class_predictions = torch.argmax(
        final_three_class_probabilities,
        dim=1
    )

    # 比较预测类别和正确类别
    three_class_correct = (
        final_three_class_predictions
        == three_class_targets
    )

    # 计算准确率
    three_class_accuracy = (
        three_class_correct.float().mean()
    )

print("预测类别：")
print(final_three_class_predictions)

print("\n正确类别：")
print(three_class_targets)

print("\n类别概率：")
print(final_three_class_probabilities)

print("\n准确率：", three_class_accuracy.item())
```

```python
# =========================
# 1. 导入需要的工具
# =========================

# 导入 PyTorch
import torch

# 导入神经网络工具
from torch import nn

# 导入图片转换工具
from torchvision import transforms

# 导入图片创建工具
from PIL import Image, ImageDraw

# 导入绘图工具
import matplotlib.pyplot as plt

# 固定随机数，使卷积层参数尽量保持一致
torch.manual_seed(42)


# =========================
# 2. 创建一张示例图片
# =========================

# 创建一张宽 300、高 200 的 RGB 图片
sample_image = Image.new(
    mode="RGB",
    size=(300, 200),
    color=(180, 220, 255)
)

# 创建绘图工具
drawer = ImageDraw.Draw(sample_image)

# 画绿色区域，模拟草地
drawer.rectangle(
    [0, 130, 300, 200],
    fill=(80, 180, 90)
)

# 画黄色圆形，模拟太阳
drawer.ellipse(
    [220, 20, 270, 70],
    fill=(255, 220, 50)
)


# =========================
# 3. 调整尺寸并转换成 Tensor
# =========================

# 创建图片处理流程
image_transform = transforms.Compose([
    # 把图片调整成高 64、宽 64
    transforms.Resize((64, 64)),

    # 把 PIL 图片转换成 Tensor
    transforms.ToTensor()
])

# 处理图片
resized_tensor = image_transform(sample_image)

print("单张图片形状：", resized_tensor.shape)


# =========================
# 4. 创建三张不同亮度的图片
# =========================

# 第一张：原始亮度
image_1 = resized_tensor

# 第二张：降低亮度
image_2 = resized_tensor * 0.5

# 第三张：提高亮度，并把数值限制在 0～1
image_3 = torch.clamp(
    resized_tensor * 1.5,
    min=0.0,
    max=1.0
)

# 把三张图片组成一个批次
three_image_batch = torch.stack(
    [image_1, image_2, image_3],
    dim=0
)

print("图片批次形状：", three_image_batch.shape)


# =========================
# 5. 创建卷积层
# =========================

# 输入有 3 个 RGB 通道
# 输出 4 张特征图
# 使用 3×3 卷积核
# padding=1 保持高度和宽度不变
conv_layer = nn.Conv2d(
    in_channels=3,
    out_channels=4,
    kernel_size=3,
    padding=1
)

# 让图片经过卷积层
conv_output = conv_layer(three_image_batch)

print("卷积输出形状：", conv_output.shape)


# =========================
# 6. 使用 ReLU
# =========================

# 创建 ReLU 激活函数
relu_layer = nn.ReLU()

# 把卷积结果中的负数变成 0
relu_output = relu_layer(conv_output)

print("ReLU 输出形状：", relu_output.shape)


# =========================
# 7. 使用最大池化
# =========================

# 每个 2×2 区域保留一个最大值
max_pool_layer = nn.MaxPool2d(
    kernel_size=2
)

# 压缩特征图的高度和宽度
pooled_output = max_pool_layer(relu_output)

print("池化输出形状：", pooled_output.shape)
```

```python
# =========================
# 8. 组合 CNN 特征提取模块
# =========================

# nn.Sequential 会按照从上到下的顺序执行各层
feature_extractor = nn.Sequential(
    # 第一步：提取局部特征
    nn.Conv2d(
        in_channels=3,
        out_channels=4,
        kernel_size=3,
        padding=1
    ),

    # 第二步：把负数变成 0，加入非线性能力
    nn.ReLU(),

    # 第三步：把高度和宽度缩小一半
    nn.MaxPool2d(
        kernel_size=2
    )
)

print(feature_extractor)
```

```python
# =========================
# 9. 使用特征提取模块
# =========================

# 把三张图片交给整个模块
features = feature_extractor(
    three_image_batch
)

print("输入形状：", three_image_batch.shape)
print("输出特征形状：", features.shape)
```

```python
# =========================
# 10. 创建 Flatten 层
# =========================

# start_dim=1 表示：
# 保留第 0 维的图片数量
# 从第 1 维开始，把后面的维度全部展开
flatten_layer = nn.Flatten(
    start_dim=1
)

# 把卷积特征展开
flattened_features = flatten_layer(
    features
)

print("展开之前：", features.shape)
print("展开之后：", flattened_features.shape)
```

```python
# =========================
# 11. 创建图片分类层
# =========================

# 每张图片输入 4096 个特征
# 最终输出 3 个类别分数
classification_layer = nn.Linear(
    in_features=4096,
    out_features=3
)

# 把展开后的图片特征交给分类层
image_scores = classification_layer(
    flattened_features
)

print("分类层输入形状：", flattened_features.shape)
print("类别分数形状：", image_scores.shape)

print("\n类别分数：")
print(image_scores)
```

```python
# =========================
# 12. 把类别分数转换成概率
# =========================

# dim=1 表示：
# 分别转换每张图片这一行的三个类别分数
image_probabilities = torch.softmax(
    image_scores,
    dim=1
)

# 选择每张图片中概率最高的类别
image_predictions = torch.argmax(
    image_probabilities,
    dim=1
)

print("类别概率：")
print(image_probabilities)

print("\n预测类别：")
print(image_predictions)

# 检查每张图片的三个类别概率之和
print("\n每张图片的概率之和：")
print(image_probabilities.sum(dim=1))
```

```python
# =========================
# 13. 创建完整的 CNN 模型
# =========================

cnn_model = nn.Sequential(
    # 第一层：从 RGB 图片中提取 4 组局部特征
    # 输入：[N, 3, 64, 64]
    # 输出：[N, 4, 64, 64]
    nn.Conv2d(
        in_channels=3,
        out_channels=4,
        kernel_size=3,
        padding=1
    ),

    # 第二层：加入非线性能力
    # 形状保持不变
    nn.ReLU(),

    # 第三层：把高度和宽度各缩小一半
    # 输出：[N, 4, 32, 32]
    nn.MaxPool2d(
        kernel_size=2
    ),

    # 第四层：保留批次维度
    # 把每张图片的特征展开
    # 输出：[N, 4096]
    nn.Flatten(
        start_dim=1
    ),

    # 第五层：根据 4096 个特征输出 3 个类别分数
    # 输出：[N, 3]
    nn.Linear(
        in_features=4 * 32 * 32,
        out_features=3
    )
)

print(cnn_model)
```

```python
# =========================
# 14. 测试完整 CNN
# =========================

# 把三张图片交给完整 CNN
cnn_scores = cnn_model(
    three_image_batch
)

# 把类别分数转换成概率
cnn_probabilities = torch.softmax(
    cnn_scores,
    dim=1
)

# 选择每张图片概率最高的类别
cnn_predictions = torch.argmax(
    cnn_probabilities,
    dim=1
)

print("输入图片形状：", three_image_batch.shape)
print("模型输出形状：", cnn_scores.shape)

print("\n类别分数：")
print(cnn_scores)

print("\n类别概率：")
print(cnn_probabilities)

print("\n预测类别：")
print(cnn_predictions)
```

```python
# =========================
# 25. 创建分批读取器
# =========================

training_loader = DataLoader(
    training_dataset,

    # 每次读取 16 张图片
    batch_size=16,

    # 每轮训练前打乱图片顺序
    shuffle=True
)

print("每一批最多包含 16 张图片")
print("一轮一共有多少批：", len(training_loader))
```

```python
# =========================
# 26. 查看 DataLoader 产生的批次
# =========================

for batch_index, (batch_images, batch_labels) in enumerate(
    training_loader
):
    print(
        "批次：",
        batch_index + 1,
        "图片形状：",
        batch_images.shape,
        "标签形状：",
        batch_labels.shape
    )
```

```python
# =========================
# 27. 创建新的 CNN
# =========================

# 固定随机数
torch.manual_seed(42)

batch_cnn_model = nn.Sequential(
    # 提取 4 组局部特征
    nn.Conv2d(
        in_channels=3,
        out_channels=4,
        kernel_size=3,
        padding=1
    ),

    # 加入非线性能力
    nn.ReLU(),

    # 把 64×64 缩小为 32×32
    nn.MaxPool2d(
        kernel_size=2
    ),

    # 把 [4, 32, 32] 展开为 4096 个特征
    nn.Flatten(
        start_dim=1
    ),

    # 输出三个类别分数
    nn.Linear(
        in_features=4 * 32 * 32,
        out_features=3
    )
)


# =========================
# 28. 创建损失函数和优化器
# =========================

# 多分类损失函数
batch_loss_function = nn.CrossEntropyLoss()

# 更新 CNN 的全部参数
batch_optimizer = torch.optim.SGD(
    batch_cnn_model.parameters(),
    lr=0.05
)
```

```python
# =========================
# 29. 使用 DataLoader 分批训练
# =========================

# 切换到训练模式
batch_cnn_model.train()

# 训练 10 轮
for epoch in range(10):
    # 记录当前一轮所有批次的总 loss
    epoch_loss = 0.0

    # 记录当前一轮预测正确的图片数量
    epoch_correct = 0

    # 记录当前一轮处理的图片总数
    epoch_total = 0

    # DataLoader 每次返回一批图片和对应标签
    for batch_images, batch_labels in training_loader:
        # 第一步：清除上一批留下的梯度
        batch_optimizer.zero_grad()

        # 第二步：计算当前批次的类别分数
        batch_scores = batch_cnn_model(
            batch_images
        )

        # 第三步：计算当前批次的 loss
        batch_loss = batch_loss_function(
            batch_scores,
            batch_labels
        )

        # 第四步：反向传播
        batch_loss.backward()

        # 第五步：更新模型参数
        batch_optimizer.step()

        # 累加当前批次的 loss
        # 乘以图片数量，得到这一批的总误差
        epoch_loss += (
            batch_loss.item() * batch_images.size(0)
        )

        # 选择每张图片分数最高的类别
        batch_predictions = torch.argmax(
            batch_scores,
            dim=1
        )

        # 统计当前批次预测正确的数量
        epoch_correct += (
            batch_predictions == batch_labels
        ).sum().item()

        # 累加已经处理的图片数量
        epoch_total += batch_images.size(0)

    # 计算这一轮中每张图片的平均 loss
    average_epoch_loss = epoch_loss / epoch_total

    # 计算这一轮的训练准确率
    epoch_accuracy = epoch_correct / epoch_total

    print(
        f"第 {epoch + 1} 轮，"
        f"平均 loss：{average_epoch_loss:.4f}，"
        f"准确率：{epoch_accuracy:.2%}"
    )
```

```python
# =========================
# 4. 定义场景图片生成函数
# =========================

def create_scene_image(scene_name, variation):
    """
    根据场景名称创建一张 64×64 的模拟图片。

    scene_name：
        city   → 城市场景
        indoor → 室内场景
        nature → 自然场景

    variation：
        用于让同一类别的图片稍微有所不同
    """

    # 创建绘图位置偏移，使图片之间不完全相同
    offset = variation % 6

    if scene_name == "city":
        # 创建蓝灰色天空
        image = Image.new(
            "RGB",
            (64, 64),
            (150, 190, 220)
        )

        drawer = ImageDraw.Draw(image)

        # 绘制灰色地面
        drawer.rectangle(
            [0, 48, 64, 64],
            fill=(90, 90, 90)
        )

        # 绘制三栋建筑
        drawer.rectangle(
            [5 + offset, 20, 20 + offset, 48],
            fill=(80, 90, 110)
        )

        drawer.rectangle(
            [24, 10 + offset, 40, 48],
            fill=(100, 100, 120)
        )

        drawer.rectangle(
            [44 - offset, 25, 60 - offset, 48],
            fill=(70, 80, 100)
        )

        # 绘制黄色窗户
        drawer.rectangle(
            [10 + offset, 27, 14 + offset, 32],
            fill=(255, 220, 80)
        )

    elif scene_name == "indoor":
        # 创建浅色墙面
        image = Image.new(
            "RGB",
            (64, 64),
            (235, 215, 185)
        )

        drawer = ImageDraw.Draw(image)

        # 绘制棕色地板
        drawer.rectangle(
            [0, 45, 64, 64],
            fill=(150, 100, 65)
        )

        # 绘制窗户
        drawer.rectangle(
            [8 + offset, 10, 28 + offset, 32],
            fill=(120, 180, 220)
        )

        # 绘制桌面
        drawer.rectangle(
            [32 - offset, 35, 58 - offset, 41],
            fill=(110, 70, 40)
        )

        # 绘制桌腿
        drawer.rectangle(
            [36 - offset, 41, 40 - offset, 55],
            fill=(90, 55, 30)
        )

        drawer.rectangle(
            [51 - offset, 41, 55 - offset, 55],
            fill=(90, 55, 30)
        )

    else:
        # 创建蓝色天空
        image = Image.new(
            "RGB",
            (64, 64),
            (130, 200, 245)
        )

        drawer = ImageDraw.Draw(image)

        # 绘制绿色草地
        drawer.rectangle(
            [0, 38, 64, 64],
            fill=(70, 175, 80)
        )

        # 绘制树干
        drawer.rectangle(
            [29 + offset, 26, 35 + offset, 51],
            fill=(105, 65, 35)
        )

        # 绘制树冠
        drawer.ellipse(
            [18 + offset, 10, 46 + offset, 38],
            fill=(35, 135, 55)
        )

        # 绘制太阳
        drawer.ellipse(
            [5, 5 + offset, 16, 16 + offset],
            fill=(255, 220, 60)
        )

    return image
```

```python
# =========================
# 9. 使用验证集检查模型
# =========================

# 切换到评估模式
scene_model.eval()

# 记录验证集累计 loss
validation_loss_total = 0.0

# 记录预测正确的图片数量
validation_correct = 0

# 记录验证图片总数
validation_total = 0

# 验证阶段不需要建立计算图和计算梯度
with torch.no_grad():
    # 分批读取验证图片
    for validation_images, validation_labels in validation_loader:
        # 使用当前模型计算类别分数
        validation_scores = scene_model(
            validation_images
        )

        # 计算当前批次的验证 loss
        validation_loss = scene_loss_function(
            validation_scores,
            validation_labels
        )

        # 当前批次有多少张图片
        current_batch_size = validation_images.size(0)

        # 累加当前批次的 loss
        validation_loss_total += (
            validation_loss.item()
            * current_batch_size
        )

        # 选择每张图片分数最高的类别
        validation_predictions = torch.argmax(
            validation_scores,
            dim=1
        )

        # 累加预测正确的数量
        validation_correct += (
            validation_predictions
            == validation_labels
        ).sum().item()

        # 累加已经验证的图片数量
        validation_total += current_batch_size


# =========================
# 10. 计算验证结果
# =========================

# 计算每张验证图片的平均 loss
validation_average_loss = (
    validation_loss_total
    / validation_total
)

# 计算验证准确率
validation_accuracy = (
    validation_correct
    / validation_total
)

print("验证图片数量：", validation_total)
print("验证平均 loss：", validation_average_loss)
print("验证准确率：", validation_accuracy)
```

```python
# =========================
# 11. 创建新的完整训练模型
# =========================

# 固定随机数
torch.manual_seed(42)

complete_model = nn.Sequential(
    # 提取 8 组图片特征
    nn.Conv2d(
        in_channels=3,
        out_channels=8,
        kernel_size=3,
        padding=1
    ),

    # 加入非线性能力
    nn.ReLU(),

    # 把 64×64 缩小成 32×32
    nn.MaxPool2d(
        kernel_size=2
    ),

    # 把每张图片的特征展开
    nn.Flatten(
        start_dim=1
    ),

    # 输出三个类别分数
    nn.Linear(
        in_features=8 * 32 * 32,
        out_features=number_of_classes
    )
)

# 创建损失函数
complete_loss_function = nn.CrossEntropyLoss()

# 创建优化器
complete_optimizer = torch.optim.SGD(
    complete_model.parameters(),
    lr=0.05
)
```

```python
# =========================
# 12. 训练并验证 10 轮
# =========================

# 保存每一轮的结果
train_loss_history = []
train_accuracy_history = []

validation_loss_history = []
validation_accuracy_history = []

# 一共训练 10 轮
for epoch in range(10):

    # =====================
    # 训练阶段
    # =====================

    # 切换到训练模式
    complete_model.train()

    # 清空这一轮的统计数据
    train_loss_total = 0.0
    train_correct = 0
    train_total = 0

    # 分批读取训练数据
    for train_images, train_labels in train_loader:
        # 清除上一批留下的梯度
        complete_optimizer.zero_grad()

        # 计算当前批次的类别分数
        train_scores = complete_model(
            train_images
        )

        # 计算当前批次的 loss
        train_loss = complete_loss_function(
            train_scores,
            train_labels
        )

        # 反向传播
        train_loss.backward()

        # 更新模型参数
        complete_optimizer.step()

        # 当前批次的图片数量
        current_batch_size = train_images.size(0)

        # 累加 loss
        train_loss_total += (
            train_loss.item()
            * current_batch_size
        )

        # 得到预测类别
        train_predictions = torch.argmax(
            train_scores,
            dim=1
        )

        # 累加预测正确的数量
        train_correct += (
            train_predictions == train_labels
        ).sum().item()

        # 累加图片总数
        train_total += current_batch_size

    # 计算这一轮的训练结果
    train_average_loss = (
        train_loss_total / train_total
    )

    train_accuracy = (
        train_correct / train_total
    )


    # =====================
    # 验证阶段
    # =====================

    # 切换到评估模式
    complete_model.eval()

    # 清空这一轮的验证统计数据
    validation_loss_total = 0.0
    validation_correct = 0
    validation_total = 0

    # 验证阶段不计算梯度
    with torch.no_grad():
        # 分批读取验证数据
        for validation_images, validation_labels in validation_loader:
            # 计算类别分数
            validation_scores = complete_model(
                validation_images
            )

            # 计算验证 loss
            validation_loss = complete_loss_function(
                validation_scores,
                validation_labels
            )

            # 当前批次图片数量
            current_batch_size = (
                validation_images.size(0)
            )

            # 累加验证 loss
            validation_loss_total += (
                validation_loss.item()
                * current_batch_size
            )

            # 得到验证集预测类别
            validation_predictions = torch.argmax(
                validation_scores,
                dim=1
            )

            # 累加预测正确的数量
            validation_correct += (
                validation_predictions
                == validation_labels
            ).sum().item()

            # 累加验证图片数量
            validation_total += current_batch_size

    # 计算这一轮的验证结果
    validation_average_loss = (
        validation_loss_total
        / validation_total
    )

    validation_accuracy = (
        validation_correct
        / validation_total
    )


    # =====================
    # 保存并显示这一轮结果
    # =====================

    train_loss_history.append(
        train_average_loss
    )

    train_accuracy_history.append(
        train_accuracy
    )

    validation_loss_history.append(
        validation_average_loss
    )

    validation_accuracy_history.append(
        validation_accuracy
    )

    print(
        f"第 {epoch + 1:2d} 轮 | "
        f"训练 loss：{train_average_loss:.4f} | "
        f"训练准确率：{train_accuracy:.2%} | "
        f"验证 loss：{validation_average_loss:.4f} | "
        f"验证准确率：{validation_accuracy:.2%}"
    )
```
